##Q2. Calculate the total marks obtained by each student across all subjects.

```python
print("Total Marks")

subjects = ["Python", "SQL", "Mathematics", "Data_Analysis"]

df["Total_Marks"] = df[subjects].sum(axis=1)

print(df[["Student_ID", "Name", "Total_Marks"]].head())
```

##Q3. Calculate Average Marks

```python
print("Average Marks")

df["Average_Marks"] = df[subjects].mean(axis=1)

print(df[["Student_ID", "Name", "Average_Marks"]].head())
```

##Q4. Assign Grades

```python
print("Grade Assignment")

def assign_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average_Marks"].apply(assign_grade)

print(df[["Student_ID", "Name", "Average_Marks", "Grade"]].head())
```

##Q.5 Identify whether each student has Passed or Failed based on their subject marks.

```python
print("Pass/Fail Status")

def check_result(row):
    if all(row[subjects] >= 40):
        return "Pass"
    else:
        return "Fail"

df["Pass_Fail"] = df.apply(check_result, axis=1)

print(df[["Student_ID", "Name", "Average_Marks", "Grade", "Pass_Fail"]].head())
```

##Q6. Calculate Overall Class Average


```python
print("Overall Class Average")

class_average = df["Average_Marks"].mean()

print("Overall Class Average:", round(class_average, 2))
```

##Q7. Identify the Highest and Lowest Performing Students

```python
print("Highest and Lowest Performing Students")

highest_student = df.loc[df["Average_Marks"].idxmax()]
lowest_student = df.loc[df["Average_Marks"].idxmin()]

print("\nHighest Performing Student")
print(highest_student[["Student_ID", "Name", "Average_Marks", "Grade"]])

print("\nLowest Performing Student")
print(lowest_student[["Student_ID", "Name", "Average_Marks", "Grade"]])
```

##Q8. Analyze Subject-wise Performance

```python
print("Subject-wise Performance")

subject_average = df[subjects].mean().round(2)

print(subject_average)
```

```python
best_subject = subject_average.idxmax()
best_average = subject_average.max()

print("\nBest Performing Subject")
print("Subject:", best_subject)
print("Average Marks:", best_average)
```

##Q9. Calculate Pass/Fail Statistics

```python
print("Pass/Fail Statistics")

result_counts = df["Pass_Fail"].value_counts()

total_students = len(df)

pass_count = result_counts.get("Pass", 0)
fail_count = result_counts.get("Fail", 0)

pass_percentage = (pass_count / total_students) * 100
fail_percentage = (fail_count / total_students) * 100

print("Total Students:", total_students)
print("Passed Students:", pass_count)
print("Failed Students:", fail_count)

print("Pass Percentage:", round(pass_percentage, 2), "%")
print("Fail Percentage:", round(fail_percentage, 2), "%")
```

##Q10. Identify the Top 5 Performing Students


```python
print("Top 5 Performing Students")

top_5_students = df.sort_values(
    by="Average_Marks",
    ascending=False
).head(5)

print(
    top_5_students[
        ["Student_ID", "Name", "Department", "Average_Marks", "Grade", "Pass_Fail"]
    ]
)
```

##Q12. Analyze the Relationship Between Attendance and Performance.

```python
print("Attendance vs Performance")

attendance_performance = df[["Attendance", "Average_Marks"]]

print(attendance_performance.head())

correlation = df["Attendance"].corr(df["Average_Marks"])

print("\nCorrelation between Attendance and Average Marks:", round(correlation, 2))
```

##Q13. Generate Final Performance Summary

```python
print("Final Performance Summary")

total_students = len(df)

class_average = df["Average_Marks"].mean()

highest_student = df.loc[df["Average_Marks"].idxmax()]
lowest_student = df.loc[df["Average_Marks"].idxmin()]

pass_count = (df["Pass_Fail"] == "Pass").sum()
fail_count = (df["Pass_Fail"] == "Fail").sum()

pass_percentage = (pass_count / total_students) * 100
fail_percentage = (fail_count / total_students) * 100

best_subject = subject_average.idxmax()
best_subject_average = subject_average.max()

attendance_correlation = df["Attendance"].corr(df["Average_Marks"])

print("Total Students:", total_students)
print("Overall Class Average:", round(class_average, 2))

print("\nHighest Performing Student:")
print(highest_student["Name"], "-", round(highest_student["Average_Marks"], 2))

print("\nLowest Performing Student:")
print(lowest_student["Name"], "-", round(lowest_student["Average_Marks"], 2))

print("\nPass/Fail Statistics:")
print("Passed:", pass_count)
print("Failed:", fail_count)
print("Pass Percentage:", round(pass_percentage, 2), "%")
print("Fail Percentage:", round(fail_percentage, 2), "%")

print("\nBest-Performing Subject:")
print(best_subject, "-", best_subject_average)

print("\nAttendance vs Performance:")
print("Correlation:", round(attendance_correlation, 2))
```
